```python
# ============================================
# ВИРТУАЛЬНЫЙ АССИСТЕНТ - КОНСУЛЬТАНТ ПО ЗДОРОВОМУ ПИТАНИЮ
# Тема: советы по правильному питанию и здоровому образу жизни
# ============================================

# ШАГ 1: Установка библиотек
!pip install --upgrade transformers
!pip install llama-cpp-python --extra-index-url https://abetlen.github.io/llama-cpp-python/whl/cpu
!pip install huggingface_hub

import warnings
warnings.filterwarnings('ignore')

print("✅ Библиотеки установлены")
print("="*60)

# ============================================
# МОДЕЛЬ 1: TRANSFORMERS (легкая диалоговая модель)
# Используем "microsoft/DialoGPT-small" - специально для чатов
# ============================================

print("\n🤖 ЗАГРУЗКА МОДЕЛИ 1 (Transformers)...")
print("Модель: microsoft/DialoGPT-small")
print("Тема: Консультант по здоровому питанию")
print("-"*60)

from transformers import AutoModelForCausalLM, AutoTokenizer
import torch

model_name_1 = "microsoft/DialoGPT-small"
tokenizer_1 = AutoTokenizer.from_pretrained(model_name_1)
model_1 = AutoModelForCausalLM.from_pretrained(model_name_1)

if tokenizer_1.pad_token is None:
    tokenizer_1.pad_token = tokenizer_1.eos_token

print("✅ Модель 1 загружена!")

# Системная роль для модели 1
SYSTEM_ROLE_1 = """Ты - профессиональный консультант по здоровому питанию.
Твои советы основаны на научных данных. Отвечай кратко, полезно и доброжелательно.
Ты не рекомендуешь экстремальные диеты и всегда учитываешь индивидуальные особенности.
Спрашивай у пользователя о его привычках питания и давай персонализированные советы."""
```

```python
# ============================================
# МОДЕЛЬ 2: LLAMA-CPP-PYTHON (легкая GGUF)
# Используем "TheBloke/TinyLlama-1.1B-Chat-v1.0-GGUF"
# ============================================

print("\n🤖 ЗАГРУЗКА МОДЕЛИ 2 (llama-cpp-python)...")
print("Модель: TheBloke/TinyLlama-1.1B-Chat-v1.0-GGUF")
print("Тема: Консультант по здоровому питанию")
print("-"*60)

from huggingface_hub import hf_hub_download
from llama_cpp import Llama

try:
    repo_id = "TheBloke/TinyLlama-1.1B-Chat-v1.0-GGUF"
    filename = "tinyllama-1.1b-chat-v1.0.Q2_K.gguf"

    print("⏳ Скачивание модели (около 400MB)...")
    model_path = hf_hub_download(repo_id=repo_id, filename=filename)

    print("⏳ Загрузка в память...")
    model_2 = Llama(
        model_path=model_path,
        n_ctx=1024,
        n_threads=2,
        n_gpu_layers=0,
        verbose=False
    )
    print("✅ Модель 2 загружена!")

    SYSTEM_ROLE_2 = """Ты - профессиональный консультант по здоровому питанию.
    Твои советы основаны на научных данных. Отвечай кратко, полезно и доброжелательно.
    Ты не рекомендуешь экстремальные диеты и всегда учитываешь индивидуальные особенности.
    Спрашивай у пользователя о его привычках питания и давай персонализированные советы."""

    model_2_loaded = True
except Exception as e:
    print(f"⚠️ Не удалось загрузить модель 2: {e}")
    print("Используем альтернативную легкую модель...")

    try:
        repo_id = "Qwen/Qwen2-0.5B-Instruct-GGUF"
        filename = "qwen2-0_5b-instruct-q4_k_m.gguf"
        model_path = hf_hub_download(repo_id=repo_id, filename=filename)
        model_2 = Llama(model_path=model_path, n_ctx=512, n_threads=2, verbose=False)
        SYSTEM_ROLE_2 = """You are a healthy eating consultant. Give brief, useful advice."""
        model_2_loaded = True
    except:
        model_2_loaded = False
        print("❌ Модель 2 не загружена, работаем только с моделью 1")

print("\n" + "="*60)
print("🍎 ВИРТУАЛЬНЫЙ АССИСТЕНТ ПО ЗДОРОВОМУ ПИТАНИЮ")
print("="*60)
```

```python
# ============================================
# ФУНКЦИИ ДЛЯ ДИАЛОГА
# ============================================

def chat_with_model_1(message, history=None):
    """Общение с моделью 1 (Transformers)"""
    if history:
        # С историей
        prompt = history + message + tokenizer_1.eos_token
    else:
        # Без истории
        prompt = SYSTEM_ROLE_1 + "\n" + message + tokenizer_1.eos_token

    inputs = tokenizer_1.encode(prompt, return_tensors='pt')
    with torch.no_grad():
        outputs = model_1.generate(
            inputs,
            max_new_tokens=150,
            pad_token_id=tokenizer_1.eos_token_id,
            do_sample=True,
            temperature=0.7,
            top_p=0.9,
            repetition_penalty=1.1
        )
    response = tokenizer_1.decode(outputs[0], skip_special_tokens=True)

    # Убираем промпт из ответа
    if response.startswith(prompt):
        response = response[len(prompt):]
    elif response.startswith(SYSTEM_ROLE_1):
        response = response[len(SYSTEM_ROLE_1):].strip()
        if response.startswith(message):
            response = response[len(message):]

    return response.strip()

def chat_with_model_2(message, history_messages=None):
    """Общение с моделью 2 (llama-cpp)"""
    if not model_2_loaded:
        return "Модель 2 не загружена"

    if history_messages:
        # С историей
        messages = history_messages
        messages.append({"role": "user", "content": message})
    else:
        # Без истории
        messages = [
            {"role": "system", "content": SYSTEM_ROLE_2},
            {"role": "user", "content": message}
        ]

    try:
        response = model_2.create_chat_completion(
            messages=messages,
            max_tokens=150,
            temperature=0.7,
            top_p=0.9
        )
        return response['choices'][0]['message']['content'].strip()
    except:
        return "Ошибка генерации"
```

```python
# ============================================
# ИНТЕРАКТИВНЫЙ РЕЖИМ С ВЫБОРОМ МОДЕЛИ
# ============================================

print("\n🎯 Выберите режим работы:")
print("1. Диалог БЕЗ сохранения истории (модель не помнит предыдущие вопросы)")
print("2. Диалог С сохранением истории (модель помнит весь разговор)")
print("\nВыберите модель:")
print("3. Transformers модель (DialoGPT-small)")
print("4. Llama-cpp модель (TinyLlama)")

choice_mode = input("\nВаш выбор (1/2/3/4): ").strip()

# Определяем режим работы
if choice_mode == "1":
    use_history = False
    print("\n🔴 РЕЖИМ: БЕЗ ИСТОРИИ - модель НЕ запоминает разговор")
    print("⚠️ Вы увидите, как модель будет давать некачественные ответы\n")
elif choice_mode == "2":
    use_history = True
    print("\n🟢 РЕЖИМ: С ИСТОРИЕЙ - модель запоминает разговор")
    print("✅ Модель будет давать связные ответы\n")
else:
    use_history = None

if choice_mode in ["1", "2"]:
    # Если выбрали только режим, спрашиваем модель
    model_choice = input("Выберите модель (3 - Transformers, 4 - Llama-cpp): ").strip()
else:
    model_choice = choice_mode

# Настройка параметров
if model_choice == "3":
    current_model = "transformers"
    print("\n🤖 Модель: Transformers (DialoGPT-small)")
elif model_choice == "4":
    if not model_2_loaded:
        print("❌ Модель 2 не загружена. Используем модель 1.")
        current_model = "transformers"
    else:
        current_model = "llama-cpp"
        print("\n🤖 Модель: Llama-cpp (TinyLlama)")
else:
    print("❌ Неверный выбор. Используем Transformers модель.")
    current_model = "transformers"
    use_history = False

# Инициализация истории
history_text = ""
history_messages = []

if use_history and current_model == "transformers":
    history_text = SYSTEM_ROLE_1 + tokenizer_1.eos_token
elif use_history and current_model == "llama-cpp":
    history_messages = [{"role": "system", "content": SYSTEM_ROLE_2}]

# ============================================
# ОСНОВНОЙ ЦИКЛ ДИАЛОГА
# ============================================

print("\n" + "="*60)
print("💬 НАЧАЛО ДИАЛОГА")
print("Введите 'exit' или 'quit' для выхода")
print("Введите 'clear' для очистки истории")
print("="*60 + "\n")

# Пример вопроса для старта
if not use_history:
    print("💡 СОВЕТ: Попробуйте задать вопрос, а затем уточняющий вопрос без контекста")
    print("   Например: 'Что полезно для завтрака?' а потом 'А что насчет обеда?'\n")

while True:
    try:
        user_input = input("🍎 Вы: ").strip()

        if user_input.lower() in ['exit', 'quit']:
            print("\n👋 До свидания! Будьте здоровы!")
            break
        elif user_input.lower() == 'clear':
            if use_history:
                if current_model == "transformers":
                    history_text = SYSTEM_ROLE_1 + tokenizer_1.eos_token
                else:
                    history_messages = [{"role": "system", "content": SYSTEM_ROLE_2}]
                print("✅ История очищена!\n")
            else:
                print("ℹ️ В режиме без истории очистка не нужна\n")
            continue
        elif user_input == "":
            continue

        # Генерация ответа
        if current_model == "transformers":
            if use_history:
                # С историей
                response = chat_with_model_1(user_input, history_text)
                history_text += user_input + tokenizer_1.eos_token + response + tokenizer_1.eos_token
            else:
                # Без истории
                response = chat_with_model_1(user_input)
        else:
            # llama-cpp модель
            if use_history:
                # С историей
                response = chat_with_model_2(user_input, history_messages)
                history_messages.append({"role": "user", "content": user_input})
                history_messages.append({"role": "assistant", "content": response})
            else:
                # Без истории
                response = chat_with_model_2(user_input)

        print(f"🤖 Ассистент: {response}\n")

    except KeyboardInterrupt:
        print("\n\n👋 Диалог прерван. До свидания!")
        break
    except Exception as e:
        print(f"❌ Ошибка: {e}\n")

# ============================================
# ДЕМОНСТРАЦИЯ ПРОБЛЕМЫ БЕЗ ИСТОРИИ
# ============================================

print("\n" + "="*60)
print("📊 ДЕМОНСТРАЦИЯ ПРОБЛЕМЫ БЕЗ СОХРАНЕНИЯ ИСТОРИИ")
print("="*60)

if current_model == "transformers" and not use_history:
    print("\n🔴 Вы сейчас общались в режиме БЕЗ истории!")
    print("Проблемы, которые вы могли заметить:")
    print("1. Модель не помнила ваше имя, если вы его называли")
    print("2. На уточняющие вопросы отвечала как на первый раз")
    print("3. Не могла продолжить логику предыдущего ответа")
    print("4. Каждый вопрос обрабатывался изолированно")

    print("\n💡 Для сравнения, вот как бы ответила модель С историей:")
    print('Вы: "Я вегетарианец"')
    print('Ассистент: "Отличный выбор! Вам стоит есть больше бобовых..."')
    print('Вы: "Что мне есть на завтрак?"')
    print('Ассистент с историей: "Как вегетарианцу, рекомендую овсянку с орехами..."')
    print('Ассистент без истории: "На завтрак можно есть яйца, каши, творог..." (не учел, что вы вегетарианец!)')

elif not use_history:
    print("\n🔴 Вы общались в режиме БЕЗ истории!")
    print("Модель не могла запомнить контекст вашего разговора.")
    print("Это приводит к нелогичным и некачественным ответам.")
else:
    print("\n🟢 Вы общались в режиме С историей!")
    print("Модель запоминала весь диалог и давала связные ответы.")
    print("\nДля сравнения, проблемы режима БЕЗ истории:")
    print("❌ Модель не помнит предыдущие вопросы")
    print("❌ Не может уточнить информацию из прошлых сообщений")
    print("❌ Каждый ответ начинается 'с чистого листа'")

print("\n" + "="*60)
print("✅ ЗАДАНИЕ ВЫПОЛНЕНО")
print("="*60)
```
